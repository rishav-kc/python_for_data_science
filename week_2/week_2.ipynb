```python
input_file = open("raw_logs.txt", "r")
master_file = open("decrypted_master.txt", "w")
alert_file = open("security_alerts.txt", "w")

for line in input_file:
    decrypted = ""

    for character in line:
        if character.isalpha():
            value = ord(character)
            value = value - 3
            decrypted = decrypted + chr(value)
        else:
            decrypted = decrypted + character

    decrypted = decrypted.strip()

    master_file.write(decrypted + "\n")

    if "BREACH" in decrypted.upper():
        alert_file.write(decrypted + "\n")

input_file.close()
master_file.close()
alert_file.close()
```
